```python
# Load Dashboard Dataset

import pandas as pd

dashboard_df = pd.read_csv(
    "executive_dashboard_dataset.csv"
)

dashboard_df.head()
```

```python
# Executive Scorecard

dashboard_df.describe()
```

```python
# Regional Performance

import matplotlib.pyplot as plt

regional_sales = (
    dashboard_df
    .groupby("Region")["Units_Sold"]
    .sum()
    .sort_values(ascending=False)
)

regional_sales.plot(kind="bar")

plt.title("Units Sold by Region")
plt.ylabel("Units Sold")
plt.show()
```

```python
# Stockout Analysis

stockouts = (
    dashboard_df
    .groupby("Region")["Stockout_Events"]
    .sum()
    .sort_values(ascending=False)
)

stockouts.plot(kind="bar", color="red")

plt.title("Stockout Events by Region")
plt.ylabel("Stockout Events")
plt.show()
```

```python
# Category Performance

category_sales = (
    dashboard_df
    .groupby("Category")["Units_Sold"]
    .sum()
    .sort_values(ascending=False)
)

category_sales.plot(kind="bar")

plt.title("Category Sales Performance")
plt.ylabel("Units Sold")
plt.show()
```

```python
# Inventory Efficiency

inventory_efficiency = (
    dashboard_df
    .groupby("Category")["Avg_Inventory"]
    .mean()
)

inventory_efficiency.plot(kind="bar")

plt.title("Average Inventory by Category")
plt.ylabel("Average Inventory")
plt.show()
```

```python
# Executibe Insights

top_region = regional_sales.idxmax()

print(f"Top Performing Region: {top_region}")

highest_stockout = stockouts.idxmax()

print(f"Highest Stockout Region: {highest_stockout}")
```

```python
## Export for Power BI

dashboard_df.to_csv(
    "powerbi_inventory_dataset.csv",
    index=False
)

print("Power BI dataset exported.")
```
