```python
## Case Study #1

case_1 = """
Scenario:

Inventory increased 15%
Sales increased 2%

What would you investigate?
"""

print(case_1)
```

```python
## Sample Strong Answer

answer = """

First I would determine whether the inventory increase
is company-wide or concentrated in specific regions,
categories, suppliers, or items.

Key areas:

1. Forecast Accuracy
2. Demand Changes
3. Supply Disruptions
4. Order Parameters
5. Seasonal Inventory Builds

Metrics:

- Weeks of Supply
- Sell Through
- Forecast Error
- Inventory Turnover
- In Stock %

Recommendation:

Correct forecast bias and adjust replenishment
targets for overstocked items.

"""

print(answer)
```

```python
## Case Study #2

case_2 = """
Forecast = 1,000 units

Actual Sales = 800 units

Questions:

1. Forecast Error
2. Forecast Accuracy
3. Business Impact
"""
print(case_2)
```

```python
## Solve Case #2

forecast = 1000
actual = 800

error = forecast - actual
error_pct = error / actual

print("Forecast Error:", error)
print("Percent Error:", round(error_pct*100,2),"%")
```
