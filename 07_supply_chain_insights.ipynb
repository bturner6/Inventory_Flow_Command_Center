# Supply Chain Insights and Recommendations
### Objective: Translate inventory and sales analytics into actionable business recommendations for leadership.

### Audience:

### Inventory Flow Leadership
### Supply Chain Directors
### Replenishment Teams
### Store Operations

```python
# Load Dashboard Data

import pandas as pd

dashboard_df = pd.read_csv(
    "executive_dashboard_dataset.csv"
)

dashboard_df.head()
```

```python
## Identify High-Risk Regions

high_risk_regions = (
    dashboard_df
    .sort_values(
        by="Stockout_Events",
        ascending=False
    )
)

high_risk_regions.head(10)
```

```python
## Inventory Imbalance Analysis

dashboard_df["Inventory_to_Sales_Ratio"] = (
    dashboard_df["Avg_Inventory"]
    / dashboard_df["Units_Sold"]
)

dashboard_df.sort_values(
    by="Inventory_to_Sales_Ratio",
    ascending=False
).head(10)
```

Insight:

Regions with elevated stockout rates are at risk of lost sales and reduced customer satisfaction.

Recommendation:

Increase replenishment frequency and review safety stock targets in these regions.

## Insight:

## High inventory-to-sales ratios suggest excess inventory.

## Recommendation:

## Reduce purchase orders and rebalance inventory across the network.

```python
## Understocked Categories

understocked = (
    dashboard_df
    .sort_values(
        by="Stockout_Events",
        ascending=False
    )
)

understocked[
    ["Category",
     "Stockout_Events"]
].head(10)
```

## Focus replenishment efforts on categories showing recurring stockouts.

## Potential benefits:

## Improved in-stock %
## Higher sales capture
## Better customer experience

```python
## Regional Opportunity Analysis

region_summary = (
    dashboard_df
    .groupby("Region")
    .agg({
        "Units_Sold":"sum",
        "Avg_Inventory":"mean",
        "Stockout_Events":"sum"
    })
)

region_summary
```

## Sales High + Inventory Low → Increase allocation

## Sales Low + Inventory High → Reduce inventory investment

## Sales High + Stockouts High → Increase safety stock

```python
## Executive Priority Matrix

dashboard_df["Priority"] = "Monitor"

dashboard_df.loc[
    dashboard_df["Stockout_Events"] > 20,
    "Priority"
] = "Immediate Action"

dashboard_df.loc[
    dashboard_df["Avg_Inventory"] >
    dashboard_df["Avg_Inventory"].quantile(.75),
    "Priority"
] = "Excess Inventory Review"

dashboard_df.head()
```

```python
##Executive Summary

print("""
EXECUTIVE INSIGHTS

1. Identify regions driving stockouts.
2. Reduce excess inventory in low-demand areas.
3. Improve replenishment frequency.
4. Focus inventory investment on high-volume categories.
5. Optimize inventory allocation by region.
""")
```
