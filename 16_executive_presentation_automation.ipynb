Executive Automation & Presentation Engine

Objectives

1. Automate KPI reporting
2. Generate executive summaries
3. Create leadership insights
4. Identify business risks
5. Build presentation-ready outputs

```python
## Imports

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from IPython.display import display
```

```python
## Load Executive Dataset

executive_df = pd.DataFrame({

    "Region":[
        "North",
        "South",
        "East",
        "West"
    ],

    "Sales":[
        400000,
        375000,
        525000,
        450000
    ],

    "Inventory":[
        200000,
        275000,
        225000,
        300000
    ],

    "Forecast_Accuracy":[
        94,
        88,
        96,
        90
    ],

    "In_Stock":[
        98,
        95,
        99,
        97
    ]
})

display(executive_df)
```

```python
## Inventory Turns

executive_df["Inventory_Turns"] = (

    executive_df["Sales"]

    /

    executive_df["Inventory"]

)

display(executive_df)
```

```python
## Weeks of Supply

executive_df["Weeks_of_Supply"] = (

    executive_df["Inventory"]

    /

    (executive_df["Sales"] / 52)

)

display(executive_df)
```

```python
## Executive KPI Summary

print("EXECUTIVE KPI SUMMARY")

print()

print(
    f"Total Sales: "
    f"${executive_df['Sales'].sum():,.0f}"
)

print(
    f"Total Inventory: "
    f"${executive_df['Inventory'].sum():,.0f}"
)

print(
    f"Average Forecast Accuracy: "
    f"{executive_df['Forecast_Accuracy'].mean():.1f}%"
)

print(
    f"Average In-Stock: "
    f"{executive_df['In_Stock'].mean():.1f}%"
)
```

```python
## Inventory Turns Visualization

plt.figure(figsize=(8,5))

sns.barplot(

    data=executive_df,

    x="Region",

    y="Inventory_Turns"

)

plt.title("Inventory Turns by Region")

plt.show()
```

```python
## Forecast Accuracy Visualization

plt.figure(figsize=(8,5))

sns.barplot(

    data=executive_df,

    x="Region",

    y="Forecast_Accuracy"

)

plt.title("Forecast Accuracy")

plt.show()
```

```python
## In-Stock Visualization

plt.figure(figsize=(8,5))

sns.barplot(

    data=executive_df,

    x="Region",

    y="In_Stock"

)

plt.title("In Stock %")

plt.show()
```

```python
## Identify Weakest Region

weakest_region = executive_df.loc[
    executive_df["Forecast_Accuracy"].idxmin()
]

weakest_region
```

```python
## Risk Detection

executive_df["Risk"] = np.where(

    executive_df["Forecast_Accuracy"] < 90,

    "High Risk",

    "Normal"

)

display(executive_df)
```

```python
## Generate Automated Insights

for _, row in executive_df.iterrows():

    if row["Forecast_Accuracy"] < 90:

        print(
            f"{row['Region']} requires "
            f"forecast improvement."
        )

    if row["In_Stock"] < 97:

        print(
            f"{row['Region']} has "
            f"in-stock opportunity."
        )
```

```python
## Executive Scorecard

scorecard = executive_df[[
    "Region",
    "Sales",
    "Inventory",
    "Inventory_Turns",
    "Forecast_Accuracy",
    "In_Stock",
    "Risk"
]]

display(scorecard)
```
